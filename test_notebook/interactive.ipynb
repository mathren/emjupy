# Interactive output

Figures that draw themselves in JavaScript cannot be shown in a text
buffer.  Each shows as a line in the output; `C-c C-f` there, or a click,
opens it -- in an xwidget where this Emacs has one, in the browser
otherwise.

```python
import numpy as np
import plotly.graph_objects as go

lon, lat = np.mgrid[0:2*np.pi:120j, -np.pi/2:np.pi/2:60j]
x, y, z = np.cos(lat)*np.cos(lon), np.cos(lat)*np.sin(lon), np.sin(lat)
elevation = np.sin(3*lon)*np.cos(2*lat) + 0.5*np.sin(5*lon + 2*lat)

fig = go.Figure(go.Surface(x=x, y=y, z=z, surfacecolor=elevation,
                           colorscale="earth", showscale=False))
fig.update_layout(title="Drag to rotate",
                  scene=dict(aspectmode="data"), margin=dict(l=0, r=0, t=40, b=0))
fig.show()
```

```python
import plotly.express as px

px.scatter(x=np.random.randn(200), y=np.random.randn(200),
           title="Hover over the points").show()
```

## Widgets

A slider from `ipywidgets` driving a matplotlib figure.

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact

@interact(frequency=(1, 10))
def wave(frequency=3):
    t = np.linspace(0, 2*np.pi, 400)
    plt.figure(figsize=(5, 2.2))
    plt.plot(t, np.sin(frequency*t))
    plt.title(f"sin({frequency} t)")
    plt.show()
```
